```python
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
sns.set_style('whitegrid')
```

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

# **Joint Plot (Fare vs Age)**

```python
sns.jointplot(x='fare', y='age', data=titanic)
plt.show()
```

# **Histogram of Fare**

```python
sns.histplot(data=titanic, x='fare', bins=30, color='salmon')
plt.show()
```

# **For older Seaborn versions:**

```python
sns.distplot(titanic['fare'], bins=30, color='salmon')
plt.show()
```

# **Box Plot (Class vs Age)**

```python
sns.boxplot(
    x='class',
    y='age',
    hue='class',
    data=titanic,
    palette=['#4C72B0', '#55A868', '#E7A44C'],
    legend=False
)
plt.show()
```

# **Swarm Plot**

```python
sns.swarmplot(
    x='class',
    y='age',
    hue='class',
    data=titanic,
    palette=['#66C2A5', '#FC8D62', '#8DA0CB'],
    dodge=False,
    legend=False
)

plt.show()
```

# **Count Plot (Male/Female)**

```python
sns.countplot(
    x='sex',
    hue='sex',
    data=titanic,
    palette=['#4C72B0', '#55A868'],
    legend=False
)

plt.show()
```

# **Heatmap (Correlation Matrix)**

```python
sns.heatmap(
    titanic.corr(numeric_only=True),
    cmap='coolwarm'
)

plt.show()
```

# **FacetGrid Histogram**

```python
g = sns.FacetGrid(titanic, col='sex')

g.map_dataframe(
    sns.histplot,
    x='age',
    bins=10
)

plt.show()
```
